```python

#  Q1. Create a 1D NumPy Array: 
# ------------------------------
#  Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values: [22.5, 25.3, 20.8, 23.4, 26.1,
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# Q2. Inspection and Properties: 
# ------------------------------
# ● Inspect the shape, data type, and the number of elements in the array.Inspect the shape of the temperatures_w1 array.

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# Q3. Array Operations:
# ----------------------
# ● Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32
# ● Find the maximum, minimum, and mean temperatures for the week.


# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

# Find Maximum, Minimum and Mean Temperature

maximum_temp = np.max(temperatures_w1)
minimum_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature:", maximum_temp)
print("Minimum Temperature:", minimum_temp)
print("Mean Temperature:", mean_temp)
```

```python
# Q4. Array Slicing and Indexing:
# --------------------------------
# ● Extract the temperatures for the first three days of the week.
# ● Extract the temperatures for the weekend (last two days).
# ● Extract the temperatures for the middle three days of the week.


# First Three Days

first_three_days = temperatures_w1[:3]

print("First Three Days:")
print(first_three_days)

# Weekend — Last Two Days

weekend = temperatures_w1[-2:]

print("Weekend:")
print(weekend)

# Middle Three Days

middle_three_days = temperatures_w1[2:5]

print("Middle Three Days:")
print(middle_three_days)
```

```python
# Q5. Create a 2D Array:
# ----------------------
# Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
# ★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
# ★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("2D Temperature Array:")
print(temperatures)
```

```python
# Q6. Inspect and Slice the 2D Array:
# --------------------------------
# ● Inspect and print the shape, data type, and total number of elements in the temperatures array.
# ● Extract the temperatures for each week and the weekends (last two days) for both weeks.


#Inspect Shape, Data Type and Total Number of Elements

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)


#Extract Temperatures for Each Week

week1 = temperatures[0]
week2 = temperatures[1]

print("Week 1 Temperatures:")
print(week1)

print("\nWeek 2 Temperatures:")
print(week2)

#Extract Weekend Temperatures for Both Weeks

week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]

print("Week 1 Weekend:")
print(week1_weekend)

print("\nWeek 2 Weekend:")
print(week2_weekend)
```

```python
# ===================
#   Pandas Series:
# =================


# Q1. Creating Pandas Series:
# ● Create a Pandas Series named marks with the following data:
# ➢ Marks: 95, 92, 89, 85, 80
# ➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("Marks Series:")
print(marks)
```

```python
# Q2. Indexing and Slicing:
# ------------------------
# ● Use the integer index position to access the mark of the 1st rank student.
# ● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
# ● Use the iloc accessor to retrieve the mark of the 3rd rank student.
# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

#Integer Index Position — 1st Rank

first_rank = marks.iloc[0]

print("Mark of 1st Rank Student:", first_rank)

#loc — Top 3 Ranks

top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print("Top 3 Ranks:")
print(top_three)

#iloc — 3rd Rank

third_rank = marks.iloc[2]

print("Mark of 3rd Rank Student:", third_rank)

#Boolean Mask — Marks Greater Than 90

marks_above_90 = marks[marks > 90]

print("Ranks with Marks Greater Than 90:")
print(marks_above_90)
```

```python
# Q3. Manipulating Series:
# --------------------------
# ● Modify the mark of the student with the 1st rank to 100.
# ● Remove the entry corresponding to the last rank from the Series.
# ● Compute the CGPA by dividing each mark in the Series by 10.
#Modify 1st Rank Mark to 100

marks.loc['Rank1'] = 100

print("Updated Marks Series:")
print(marks)

#Remove Last Rank

marks = marks.drop('Rank5')

print("Series After Removing Last Rank:")
print(marks)
#Calculate CGPA

cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
# ======================
#   Pandas DataFrame:
# ======================

# Q1. Creating Pandas DataFrame:
# ● Create a Pandas DataFrame named transactions with the following data:
# ➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
# ➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
# ➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'
# ➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
# Q2. Data Exploration:
# ----------------------
# ● Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.
# ● Display only the 'ProductCategory' and 'Amount' columns.
# ● Retrieve the last 3 columns of the DataFrame using iloc or loc.
# ● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
# ● Find the value counts for the 'ProductCategory' column.
# ● Find the unique values in the 'Region' column.
# ● Group by 'Region' and find the mean amount for each region.

#Display DataFrame
print(transactions)
```

```python
#Retrieve Last 3 Columns Using iloc
last_three_columns = transactions.iloc[:, -3:]

print(last_three_columns)
```

```python
#Filter North Region and Amount Greater Than 200
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_transactions)
```

```python
#Value Counts for ProductCategory
product_counts = transactions['ProductCategory'].value_counts()

print(product_counts)
```

```python
#Unique Values in Region
unique_regions = transactions['Region'].unique()

print("Unique Regions:")
print(unique_regions)
```

```python
#Group by Region and Find Mean Amount
region_mean = transactions.groupby('Region')['Amount'].mean()

print("Mean Amount by Region:")
print(region_mean)
```

```python
# Q3. Manipulating the DataFrame:
# --------------------------------
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.
# ● Remove the row with TransactionID 109.
# ● Delete the 'Discount' column from the DataFrame.

#Modify TransactionID 102 Amount to 165
transactions.loc[
    transactions['TransactionID'] == 102,
    'Amount'
] = 165

print(transactions)
```

```python
#Add Discount Column — 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
#Remove TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

```python
#Delete Discount Column
transactions = transactions.drop(columns='Discount')

print(transactions)
```
